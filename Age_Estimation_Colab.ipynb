### Bước 3: Thiết lập môi trường và Giải nén dữ liệu
Copy thư mục `src` và giải nén `UTKFace.zip` từ Drive vào bộ nhớ cục bộ của Colab để tăng tốc độ I/O.

```python
import os
import shutil

# Đường dẫn trên Drive của bạn
DRIVE_ZIP_PATH = "/content/drive/MyDrive/UTKFace.zip"
DRIVE_SRC_DIR = "/content/drive/MyDrive/src"

# Đường dẫn làm việc cục bộ trên Colab
LOCAL_PROJECT_DIR = "/content/project"
LOCAL_SRC_DIR = os.path.join(LOCAL_PROJECT_DIR, "src")
LOCAL_DATASET_DIR = os.path.join(LOCAL_PROJECT_DIR, "dataset")
LOCAL_MODELS_DIR = os.path.join(LOCAL_PROJECT_DIR, "models")
LOCAL_RESULTS_DIR = os.path.join(LOCAL_PROJECT_DIR, "results")

# Tạo các thư mục cần thiết cục bộ
os.makedirs(LOCAL_PROJECT_DIR, exist_ok=True)
os.makedirs(LOCAL_DATASET_DIR, exist_ok=True)
os.makedirs(LOCAL_MODELS_DIR, exist_ok=True)
os.makedirs(LOCAL_RESULTS_DIR, exist_ok=True)

# Copy thư mục src từ Drive về Colab cục bộ
if os.path.exists(LOCAL_SRC_DIR):
    shutil.rmtree(LOCAL_SRC_DIR)
if os.path.exists(DRIVE_SRC_DIR):
    shutil.copytree(DRIVE_SRC_DIR, LOCAL_SRC_DIR)
    print("✓ Đã copy thư mục src về Colab cục bộ.")
else:
    print(f"❌ Không tìm thấy thư mục {DRIVE_SRC_DIR} trên Drive.")

# Giải nén Dataset UTKFace.zip
if os.path.exists(DRIVE_ZIP_PATH):
    print("Đang giải nén UTKFace.zip từ Drive vào Colab cục bộ...")
    shutil.unpack_archive(DRIVE_ZIP_PATH, LOCAL_DATASET_DIR)
    print("✓ Giải nén thành công dataset.")
else:
    print(f"❌ Không tìm thấy tệp {DRIVE_ZIP_PATH} trên Drive.")
```
